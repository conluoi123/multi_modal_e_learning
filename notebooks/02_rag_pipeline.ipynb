## Import thư viện & đường dẫn

```python
import sys
import os
import time
from IPython.display import display, Markdown, HTML

# Thêm thư mục gốc của dự án vào đường dẫn
sys.path.append(os.path.abspath('..'))

from backend.rag.retriever import retrieve_context
from backend.rag.generator import generate_answer
```

## Retrieval

```python
# Bạn thử nhập câu hỏi bất kỳ liên quan đến file PDF mẫu
query = "Nhóm cần làm mấy tiêu chí, tiêu chí báo cáo tiếng Việt bao nhiêu điểm?"

print(" Đang tìm kiếm trong ChromaDB...")
start_time = time.time()
chunks = retrive_context(query, k=3)
search_time = time.time() - start_time

# In kết quả trực quan
html_output = f"<h3>Kết quả tìm kiếm (Thời gian: {search_time:.2f}s)</h3>"
for i, chunk in enumerate(chunks):
    source = chunk['metadata'].get('source')
    page = chunk['metadata'].get('page')
    
    html_output += f"""
    <div style='border-left: 4px solid #2196F3; padding-left: 10px; margin-bottom: 15px; background-color: #f9f9f9; padding: 10px;'>
        <b> Nguồn:</b> {source} (Trang {page})<br>
        <b> Nội dung trích xuất:</b> <i>{chunk['text'][:300]}...</i>
    </div>
    """
display(HTML(html_output))
```

## Generation

```python
print(" Đang gửi ngữ cảnh cho Gemini suy nghĩ...")
start_time = time.time()
answer = generate_answer(query, chunks)
gen_time = time.time() - start_time

# Hiển thị câu trả lời bằng Markdown
display(Markdown(f"### 💡 Trả lời của AI (Mất {gen_time:.2f}s)"))
display(Markdown(answer))
display(Markdown("---"))

print("HOÀN TẤT LUỒNG RAG END-TO-END!")
```
