## Kiểm chứng kết quả ở bước Ingestion

```python
# !pip install matplotlib
import sys
import os
import matplotlib.pyplot as plt
# Thêm thư mục gốc vào đường dẫn để import được module backend
sys.path.append(os.path.abspath('..'))

from backend.ingestion.pdf_parser import parse_pdf
from backend.ingestion.chunker import chunk_text

# Đọc file PDF
pages = parse_pdf("../data/raw/sample.pdf")
print(f"Đã đọc {len(pages)} trang.")

# Cắt chunk
chunks = chunk_text(pages)
print(f"Đã cắt thành {len(chunks)} chunks.")
```

## Trực quan hóa độ dài các chunk 

```python
# Trực quan hóa độ dài các chunk bằng biểu đồ
chunk_lengths = [len(c["text"]) for c in chunks]

plt.figure(figsize=(10, 5))
plt.hist(chunk_lengths, bins=20, color='skyblue', edgecolor='black')
plt.title("Phân bổ độ dài của các Chunks (ký tự)")
plt.xlabel("Số lượng ký tự")
plt.ylabel("Số lượng Chunk")
plt.axvline(x=600, color='red', linestyle='--', label='Chunk Size Limit (600)')
plt.legend()
plt.show()
```
