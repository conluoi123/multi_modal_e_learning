Đang yêu cầu AI giải Toán bằng LaTeX...
==================================================
Chào em, với tư cách là một giáo viên Toán, thầy sẽ hướng dẫn em giải phương trình bậc hai $x^2 - 5x + 6 = 0$ một cách chi tiết nhất.

Để giải phương trình dạng tổng quát $ax^2 + bx + c = 0$ (với $a \neq 0$), chúng ta có thể sử dụng phương pháp tính biệt thức $\Delta$ (Delta).

### Bước 1: Xác định các hệ số
Từ phương trình $x^2 - 5x + 6 = 0$, ta xác định được các hệ số:
- $a = 1$
- $b = -5$
- $c = 6$

### Bước 2: Tính biệt thức $\Delta$
Công thức tính biệt thức là:
$$\Delta = b^2 - 4ac$$

Thay các giá trị vào công thức:
$$\Delta = (-5)^2 - 4 \cdot 1 \cdot 6$$
$$\Delta = 25 - 24$$
$$\Delta = 1$$

Vì $\Delta > 0$, phương trình có hai nghiệm phân biệt.

### Bước 3: Tìm nghiệm của phương trình
Công thức nghiệm là:
$$x = \frac{-b \pm \sqrt{\Delta}}{2a}$$

Thay các giá trị vào công thức:
$$x = \frac{-(-5) \pm \sqrt{1}}{2 \cdot 1}$$
$$x = \frac{5 \pm 1}{2}$$

Từ đó, ta tìm được hai nghiệm:
- Nghiệm thứ nhất: $x_1 = \frac{5 + 1}{2} = \frac{6}{2} = 3$
- Nghiệm thứ hai: $x_2 = \frac{5 - 1}{2} = \frac{4}{2} = 2$

### Kết luận
Vậy tập nghiệm của phương trình là $S = \{2; 3\}$.

---
*Mẹo nhỏ:* Đối với phương trình này, em cũng có thể sử dụng phương pháp phân tích đa thức thành nhân tử:
$$x^2 - 5x + 6 = (x - 2)(x - 3) = 0$$
Từ đó dễ dàng suy ra $x - 2 = 0$ hoặc $x - 3 = 0$, tương đương với $x = 2$ hoặc $x = 3$. Chúc em học tốt!
